```python
import requests

isbn = "9780071605878"  # example ISBN
url = f"https://www.googleapis.com/books/v1/volumes?q=isbn:{isbn}"

r = requests.get(url)
data = r.json()

print(data["items"][0]["volumeInfo"])
```

```python
import pandas as pd

df = pd.json_normalize(data["items"][0]["volumeInfo"])
print(df)
```
